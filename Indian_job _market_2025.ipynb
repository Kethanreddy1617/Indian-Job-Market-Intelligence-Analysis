```python
# Clean column names
df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

df_clean.columns.tolist()
```

```python
df_clean = df_clean.dropna(subset=['companyname'])
```

```python
df_clean['tagsandskills'] = df_clean['tagsandskills'].fillna('Not Specified')
```

```python
columns_to_fill = [
    'minimumsalary',
    'maximumsalary',
    'minimumexperience',
    'maximumexperience'
]

df_clean[columns_to_fill] = df_clean[columns_to_fill].fillna(0)
```
