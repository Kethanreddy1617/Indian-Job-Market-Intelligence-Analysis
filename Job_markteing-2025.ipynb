```python
import pandas as pd
df = pd.read_excel(r"C:\Users\91934\OneDrive\Documents\Kethan\Data Science\Projet\indian-job-market-dataset-2025.xlsx")
df_clean = df.copy()
```

```python
df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
```

```python
df_clean = df_clean.drop_duplicates()

df_clean = df_clean.dropna(subset=['companyname'])

df_clean['tagsandskills'] = df_clean['tagsandskills'].fillna('Not Specified')
df_clean['experience'] = df_clean['experience'].fillna('Not Specified')

columns_to_fill = [
    'minimumsalary',
    'maximumsalary',
    'minimumexperience',
    'maximumexperience',
    'reviewscount',
    'aggregaterating'
]

df_clean[columns_to_fill] = df_clean[columns_to_fill].fillna(0)
```

```python
text_columns = [
    'title',
    'currency',
    'companyname',
    'tagsandskills',
    'experience',
    'location',
    'jobdescription'
]

for col in text_columns:
    df_clean[col] = df_clean[col].astype(str).str.strip()
```

```python
df_clean['days_since_posted'] = (
    df_clean['jobuploaded']
    .str.extract(r'(\d+)')[0]
    .astype(float)
)
```

```python
df_clean['average_salary'] = (
    df_clean['minimumsalary'] +
    df_clean['maximumsalary']
) / 2

df_clean['average_salary'] = df_clean['average_salary'].replace(0, pd.NA)
```
